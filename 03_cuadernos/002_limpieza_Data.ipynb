```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns



# Vista rápida
df.head()
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df['Score'], bins=5, kde=True)
plt.title("Distribución de Score")
plt.show()
```

```python
df['helpfulness_ratio'] = df['HelpfulnessNumerator'] / df['HelpfulnessDenominator']
df['helpfulness_ratio'] = df['helpfulness_ratio'].fillna(0)

plt.figure(figsize=(6,4))
sns.histplot(df['helpfulness_ratio'], bins=20, kde=True)
plt.title("Distribución Helpfulness Ratio")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x=df['Score'])
plt.title("Boxplot Score")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x=df['helpfulness_ratio'])
plt.title("Boxplot Helpfulness Ratio")
plt.show()
```

```python

#Detectar outliers (IQR + Z-score)
#📌 Método IQR

Q1 = df['helpfulness_ratio'].quantile(0.25)
Q3 = df['helpfulness_ratio'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers_iqr = df[
    (df['helpfulness_ratio'] < lower_bound) |
    (df['helpfulness_ratio'] > upper_bound)
]

print("Outliers IQR:", len(outliers_iqr))
```
